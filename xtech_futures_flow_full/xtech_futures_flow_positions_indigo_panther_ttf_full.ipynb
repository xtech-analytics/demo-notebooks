```python
from unifier import unifier as Unifier
```

```python
Unifier.user = 'unifier username'
Unifier.token = 'unifier api token'
os.environ['UNIFIER_USER'] = Unifier.user
os.environ['UNIFIER_TOKEN'] = Unifier.token
dataset_name = "xtech_futures_flow_positions_indigo_panther_tff_full"
dataset_identifier = "xtech_futures_flow_ident_indigo_panther_tff_full"
```

## Futures Reference Table

```python
df =  Unifier.get_dataframe(name=dataset_identifier, asof_date="2026-11-11")
display(df)
```

## XTech Global Futures Flow Forecasts

```python
df =  Unifier.get_dataframe(name=dataset_name,  back_to='2025-11-18', limit=10)
display(df)
```

```python
df = Unifier.get_dataframe(name=dataset_name, back_to='2024-01-11', up_to='2024-02-01', limit=10)
display(df)
```

```python
df =  Unifier.get_dataframe(name=dataset_name, key="CSI_100048", back_to='2025-11-18', limit=10)
display(df)
```

```python
df = Unifier.get_dataframe(name=dataset_name, key="CSI_100048", back_to='2024-01-11', up_to='2024-02-01', limit=10)
display(df)
```

## Simply Import and initialize Unifier and Go!

```python
from unifier import unifier as Unifier
```

```python
Unifier.user = 'unifier username'
Unifier.token = 'unifier api token'
os.environ['UNIFIER_USER'] = Unifier.user
os.environ['UNIFIER_TOKEN'] = Unifier.token
```

## Retrieve data for a specific date using the asof_date parameter

```python
df = Unifier.get_dataframe(name="kalshi_macro_orderbook_levels", asof_date='2026-10-22', limit=10)
display(df)
```

## Retrieve data for a specific date range using the back_to and up_to parameters


```python
df = Unifier.get_dataframe(name='kalshi_macro_orderbook_levels', back_to='2026-10-22', up_to='2026-10-23', limit=100)
display(df)
```

## Retrieve data for a specific date with a specific ticker (key='ticker') using the asof_date parameter

```python
df = Unifier.get_dataframe(name='kalshi_macro_orderbook_levels',key='KXFED-26DEC-T4.50', asof_date='2026-10-22', limit=10)
display(df)
```

## Retrieve data for a specific date range with a specific ticker (key='ticker') using the back_to and up_to parameters


```python
df = Unifier.get_dataframe(name='kalshi_macro_orderbook_levels',key='KXFED-26DEC-T4.50', back_to='2026-10-22', up_to='2026-10-23', limit=10)
display(df)
```

## Download data file for a specific date using asof_date


```python
Unifier.replicate(name="kalshi_macro_orderbook_levels", target_location="./data", asof_date='2026-10-22')
```

## Download all data files

```python
Unifier.replicate(name="kalshi_macro_orderbook_levels", target_location="./data")
```

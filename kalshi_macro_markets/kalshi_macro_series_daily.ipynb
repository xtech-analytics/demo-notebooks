## Simply Import and initialize Unifier and Go!

```python
from unifier import unifier as Unifier
```

```python
Unifier.user = 'unifier username'
Unifier.token = 'unifier api token'
os.environ['UNIFIER_USER'] = Unifier.user
os.environ['UNIFIER_TOKEN'] = Unifier.token
```

## Retrieve data for a specific date using the asof_date parameter

```python
df = Unifier.get_dataframe(name="kalshi_macro_series_daily", asof_date='2026-10-23', limit=10)
display(df)
```

## Retrieve data for a specific date range using the back_to and up_to parameters


```python
df = Unifier.get_dataframe(name='kalshi_macro_series_daily', back_to='2026-10-19', up_to='2026-10-23', limit=10)
display(df)
```

## Retrieve data for a specific date with a specific series_ticker (key='series_ticker') using the asof_date parameter

```python
df = Unifier.get_dataframe(name='kalshi_macro_series_daily',key='KXCPIYOY', asof_date='2026-10-23', limit=10)
display(df)
```

## Retrieve data for a specific date range with a specific series_ticker (key='series_ticker') using the back_to and up_to parameters


```python
df = Unifier.get_dataframe(name='kalshi_macro_series_daily',key='KXCPIYOY', back_to='2026-10-19', up_to='2026-10-23', limit=10)
display(df)
```

## Download data file for a specific date using asof_date


```python
Unifier.replicate(name="kalshi_macro_series_daily", target_location="./data", asof_date='2026-10-22')
```

## Simply Import and initialize Unifier and Go!

```python
from unifier import unifier as Unifier
```

```python
Unifier.user = 'unifier username'
Unifier.token = 'unifier api token'
os.environ['UNIFIER_USER'] = Unifier.user
os.environ['UNIFIER_TOKEN'] = Unifier.token
```

## Retrieve data for a specific date using the asof_date parameter

```python
df = Unifier.get_dataframe(name="kalshi_macro_daily_distribution", asof_date='2026-10-21', limit=10)
display(df)
```

## Retrieve data for a specific date range using the back_to and up_to parameters


```python
df = Unifier.get_dataframe(name='kalshi_macro_daily_distribution', back_to='2026-09-01', up_to='2026-10-22', limit=10)
display(df)
```

## Retrieve data for a specific date with a specific contract_preamble (key='contract_preamble') using the asof_date parameter

```python
df = Unifier.get_dataframe(name='kalshi_macro_daily_distribution',key='KXFED-26OCT', asof_date='2026-10-15', limit=10)
display(df)
```

## Retrieve data for a specific date range with a specific contract_preamble (key='contract_preamble') using the back_to and up_to parameters


```python
df = Unifier.get_dataframe(name='kalshi_macro_daily_distribution',key='KXFED-26OCT', back_to='2026-09-01', up_to='2026-10-22', limit=10)
display(df)
```

## Download data file for a specific date using asof_date


```python
Unifier.replicate(name="kalshi_macro_daily_distribution", target_location="./data", asof_date='2026-10-15')
```

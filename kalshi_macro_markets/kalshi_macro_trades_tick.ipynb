## Simply Import and initialize Unifier and Go!

```python
from unifier import unifier as Unifier
```

```python
Unifier.user = 'unifier username'
Unifier.token = 'unifier api token'
os.environ['UNIFIER_USER'] = Unifier.user
os.environ['UNIFIER_TOKEN'] = Unifier.token
```

## Retrieve data for a specific date using the asof_date parameter

```python
df = Unifier.get_dataframe(name="kalshi_macro_trades_tick", asof_date='2026-10-22', limit=10)
display(df)
```

## Retrieve data for a specific date range using the back_to and up_to parameters


```python
df = Unifier.get_dataframe(name='kalshi_macro_trades_tick', back_to='2026-10-01', up_to='2026-10-23', limit=100)
display(df)
```

## Retrieve data for a specific date with a specific ticker (key='ticker') using the asof_date parameter

```python
df = Unifier.get_dataframe(name='kalshi_macro_trades_tick',key='KXCPI-26NOV-T-0.1', asof_date='2026-10-23', limit=10)
display(df)
```

## Retrieve data for a specific date range with a specific ticker (key='ticker') using the back_to and up_to parameters


```python
df = Unifier.get_dataframe(name='kalshi_macro_trades_tick',key='KXFED-26OCT-T4.00', back_to='2026-10-01', up_to='2026-10-23', limit=10)
display(df)
```

## Download data file for a specific date using asof_date


```python
Unifier.replicate(name="kalshi_macro_trades_tick", target_location="./data", asof_date='2026-10-15')
```

## Download all data files

```python
Unifier.replicate(name="kalshi_macro_trades_tick", target_location="./data")
```

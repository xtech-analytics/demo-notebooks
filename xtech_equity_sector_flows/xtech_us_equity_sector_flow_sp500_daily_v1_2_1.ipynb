```python
df =  Unifier.get_dataframe(name="xtech_us_equity_sector_flow_sp500_daily_v1_2_1", back_to='2025-11-18', limit=10)
display(df)
```

```python
df = Unifier.get_dataframe(name="xtech_us_equity_sector_flow_sp500_daily_v1_2_1", back_to='2024-01-11', up_to='2024-02-01', limit=10)
display(df)
```
